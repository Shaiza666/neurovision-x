# Data inspection

Research prototype — not a clinically validated diagnostic tool.

This notebook prints whatever BraTS-like tree is under `configs/data.yaml` `paths.raw_root`. It does not download data and does not synthesize MRI.

```python
from pathlib import Path
from neurovisionx.data.errors import MissingBraTSDataError
from neurovisionx.data.inventory import discover_inventory
from neurovisionx.utils.config import load_yaml
from neurovisionx.utils.paths import project_root

cfg = load_yaml("configs/data.yaml")
raw = project_root() / cfg["paths"]["raw_root"]
print("raw_root:", raw)
print("exists:", raw.exists())
try:
    inventory = discover_inventory(raw)
except MissingBraTSDataError as exc:
    print(exc)
else:
    print("layout:", inventory.layout)
    print("cases:", inventory.n_cases, "complete:", inventory.n_complete)
    print("naming:", inventory.naming_counts)
    print("notes:", inventory.notes)
    for case in inventory.cases[:10]:
        print(case.patient_id, "missing=", case.missing, "naming=", case.naming)
```
